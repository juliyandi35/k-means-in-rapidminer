```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
# Ganti path dengan lokasi file kamu
df = pd.read_excel("D:/Kerjaan/Research Consultant/Project K-Means in Rapidminer/Data/Data 2020.xlsx")
```

```python
X = df[['Billing Value', 'Billing Qty.']]

# Standarisasi data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
n_clusters = 5

kmeans = KMeans(n_clusters=n_clusters, random_state=42)
labels = kmeans.fit_predict(X_scaled)

# Tambahkan label cluster ke dataframe
df['Cluster'] = labels
```

```python
# Silhouette Score
sil_score = silhouette_score(X_scaled, labels)
print(f"Silhouette Score: {sil_score:.2f}")
```

```python
# Hitung Dominan Brand per Cluster berdasarkan frekuensi terbesar
dominant_brand = df.groupby('Cluster')['Material Group'].agg(lambda x: x.mode().iloc[0] if not x.empty else None).reset_index().rename(columns={'Material Group': 'Dominan Brand'})

# Hitung Dominan Healthy Brand per Cluster berdasarkan 5 elemen yang disebutkan
healthy_brands = ['BIOKUL', 'JUNGLE', 'DIAMOND', 'OATSIDE', 'V-SOY']
dominant_healthy_brand = df.groupby('Cluster')['Material Group'].agg(
    lambda x: x[x.isin(healthy_brands)].mode().iloc[0] if not x[x.isin(healthy_brands)].empty else None
).reset_index().rename(columns={'Material Group': 'Dominan Healthy Brand'})

# Gabungkan kedua hasil ke dalam satu dataframe
result = pd.merge(dominant_brand, dominant_healthy_brand, on='Cluster', how='left')

# Export ke file Excel
result.to_excel('D:/Kerjaan/Research Consultant/Project K-Means in Rapidminer/Cluster Dominan Brand 2020 Adjusted.xlsx', index=False)
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(data=df, x='Billing Value', y='Billing Qty.', hue='Cluster', palette='tab10')
plt.title(f"KMeans Clustering 2020 Data (K={n_clusters})\nSilhouette Score: {sil_score:.2f}")
plt.xlabel("Billing Value")
plt.ylabel("Billing Qty")
plt.legend(title="Cluster")
plt.show()
```
